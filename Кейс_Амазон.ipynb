```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Загружаем файл (замени путь на свой!)
file_path = '/content/amazon_market_data.csv'  # ← ЗАМЕНИ!

df = pd.read_csv(file_path)

print("✅ Данные Amazon загружены!")
print(f"📊 Размер: {df.shape[0]} строк, {df.shape[1]} колонок")

# Смотрим колонки
print("\n📋 Колонки:")
print(df.columns.tolist())

print("\n📊 Первые 5 строк:")
print(df.head())
```

```python
print("\n" + "="*50)
print("🔍 АНАЛИЗ ДАННЫХ AMAZON")
print("="*50)

print("\n📋 ВСЕ КОЛОНКИ:")
for i, col in enumerate(df.columns, 1):
    print(f"   {i}. {col}")

print("\n📊 ТИПЫ ДАННЫХ:")
print(df.dtypes)

print("\n📊 СТАТИСТИКА ПО ЧИСЛОВЫМ КОЛОНКАМ:")
print(df.describe())
```

```python
print("\n" + "="*50)
print("🔧 ОЧИСТКА ДАННЫХ")
print("="*50)

# 1. Пропуски
print("📊 Пропуски:")
print(df.isna().sum())

# 2. Удаляем пропуски
df = df.dropna()
print(f"✅ После удаления пропусков: {len(df)} строк")

# 3. Дубликаты
duplicates = df.duplicated().sum()
print(f"📊 Дубликатов: {duplicates}")
df = df.drop_duplicates()
print(f"✅ После удаления дубликатов: {len(df)} строк")

# 4. Преобразуем дату (если есть)
date_col = 'Order Date' if 'Order Date' in df.columns else None
if date_col and date_col in df.columns:
    df[date_col] = pd.to_datetime(df[date_col], errors='coerce')
    print(f"✅ {date_col} преобразован в datetime")
```

```python
print("\n" + "="*50)
print("📊 ОБЩАЯ СТАТИСТИКА AMAZON")
print("="*50)

# Находим нужные колонки (названия могут отличаться!)
order_col = None
customer_col = None
sales_col = None
date_col = None
category_col = None
subcategory_col = None
product_col = None
city_col = None
state_col = None
region_col = None
quantity_col = None
discount_col = None

for col in df.columns:
    col_lower = col.lower()
    if 'order' in col_lower and 'id' in col_lower:
        order_col = col
    elif 'customer' in col_lower and 'id' in col_lower:
        customer_col = col
    elif 'sales' in col_lower:
        sales_col = col
    elif 'order date' in col_lower:
        date_col = col
    elif 'category' in col_lower and 'sub' not in col_lower:
        category_col = col
    elif 'sub' in col_lower and 'category' in col_lower:
        subcategory_col = col
    elif 'product name' in col_lower:
        product_col = col
    elif 'city' in col_lower:
        city_col = col
    elif 'state' in col_lower:
        state_col = col
    elif 'region' in col_lower:
        region_col = col
    elif 'quantity' in col_lower:
        quantity_col = col
    elif 'discount' in col_lower:
        discount_col = col

print(f"📋 Найдены колонки:")
print(f"   Заказы: {order_col}")
print(f"   Клиенты: {customer_col}")
print(f"   Выручка: {sales_col}")
print(f"   Дата: {date_col}")
print(f"   Категория: {category_col}")
print(f"   Подкатегория: {subcategory_col}")
print(f"   Товар: {product_col}")
print(f"   Город: {city_col}")
print(f"   Штат: {state_col}")
print(f"   Регион: {region_col}")

if order_col:
    print(f"\n📊 Всего заказов: {df[order_col].nunique()}")
if customer_col:
    print(f"📊 Всего клиентов: {df[customer_col].nunique()}")
if sales_col:
    print(f"📊 Общая выручка: {df[sales_col].sum():,.2f}")
    print(f"📊 Средний чек: {df[sales_col].mean():.2f}")
    print(f"📊 Медианный чек: {df[sales_col].median():.2f}")
```

```python
if region_col:
    print("\n" + "="*50)
    print("📊 ЗАКАЗЫ ПО РЕГИОНАМ")
    print("="*50)

    region_counts = df[region_col].value_counts()
    region_sales = df.groupby(region_col)[sales_col].sum().sort_values(ascending=False)

    print("📊 Количество заказов по регионам:")
    print(region_counts)
    print("\n📊 Выручка по регионам:")
    print(region_sales)

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    region_counts.plot(kind='bar', ax=axes[0], color='skyblue', edgecolor='black')
    axes[0].set_title('Количество заказов по регионам', fontsize=12)
    axes[0].set_xlabel('Регион', fontsize=10)
    axes[0].set_ylabel('Количество заказов', fontsize=10)
    axes[0].tick_params(axis='x', rotation=45)

    region_sales.plot(kind='bar', ax=axes[1], color='lightgreen', edgecolor='black')
    axes[1].set_title('Выручка по регионам', fontsize=12)
    axes[1].set_xlabel('Регион', fontsize=10)
    axes[1].set_ylabel('Выручка', fontsize=10)
    axes[1].tick_params(axis='x', rotation=45)

    plt.tight_layout()
    plt.show()
```

```python
if state_col:
    print("\n" + "="*50)
    print("📊 ТОП-10 ШТАТОВ")
    print("="*50)

    state_counts = df[state_col].value_counts().head(10)
    print(state_counts)

    plt.figure(figsize=(12, 5))
    state_counts.plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title('Топ-10 штатов по количеству заказов', fontsize=14)
    plt.xlabel('Штат', fontsize=12)
    plt.ylabel('Количество заказов', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

if city_col:
    print("\n" + "="*50)
    print("📊 ТОП-10 ГОРОДОВ")
    print("="*50)

    city_counts = df[city_col].value_counts().head(10)
    print(city_counts)

    plt.figure(figsize=(12, 5))
    city_counts.plot(kind='bar', color='lightgreen', edgecolor='black')
    plt.title('Топ-10 городов по количеству заказов', fontsize=14)
    plt.xlabel('Город', fontsize=12)
    plt.ylabel('Количество заказов', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

```python
if category_col and sales_col:
    print("\n" + "="*50)
    print("📊 ПРОДАЖИ ПО КАТЕГОРИЯМ")
    print("="*50)

    category_sales = df.groupby(category_col)[sales_col].sum().sort_values(ascending=False)
    print(category_sales)

    plt.figure(figsize=(12, 5))
    category_sales.plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title('Продажи по категориям', fontsize=14)
    plt.xlabel('Категория', fontsize=12)
    plt.ylabel('Выручка', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    print(f"🏆 Самая популярная категория: {category_sales.index[0]}")

if subcategory_col and sales_col:
    print("\n" + "="*50)
    print("📊 ТОП-10 ПОДКАТЕГОРИЙ ПО ВЫРУЧКЕ")
    print("="*50)

    subcategory_sales = df.groupby(subcategory_col)[sales_col].sum().sort_values(ascending=False).head(10)
    print(subcategory_sales)

    plt.figure(figsize=(12, 5))
    subcategory_sales.plot(kind='bar', color='lightgreen', edgecolor='black')
    plt.title('Топ-10 подкатегорий по выручке', fontsize=14)
    plt.xlabel('Подкатегория', fontsize=12)
    plt.ylabel('Выручка', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

```python
if product_col and sales_col:
    print("\n" + "="*50)
    print("📊 ТОП-10 ТОВАРОВ ПО ВЫРУЧКЕ")
    print("="*50)

    top_products = df.groupby(product_col)[sales_col].sum().sort_values(ascending=False).head(10)
    print(top_products)

    plt.figure(figsize=(12, 6))
    top_products.plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title('Топ-10 товаров по выручке', fontsize=14)
    plt.xlabel('Товар', fontsize=12)
    plt.ylabel('Выручка', fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```

```python
if date_col and sales_col:
    print("\n" + "="*50)
    print("📊 ДИНАМИКА ПРОДАЖ")
    print("="*50)

    # Группируем по дате
    df[date_col] = pd.to_datetime(df[date_col], errors='coerce')

    # Ежедневная динамика
    daily_sales = df.groupby(df[date_col].dt.date)[sales_col].sum()

    plt.figure(figsize=(14, 5))
    daily_sales.plot(color='blue', linewidth=2)
    plt.title('Динамика ежедневных продаж', fontsize=14)
    plt.xlabel('Дата', fontsize=12)
    plt.ylabel('Выручка', fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.3)
    plt.tight_layout()
    plt.show()

    # Месячная динамика
    df['month'] = df[date_col].dt.to_period('M')
    monthly_sales = df.groupby('month')[sales_col].sum()

    plt.figure(figsize=(12, 5))
    monthly_sales.plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title('Динамика продаж по месяцам', fontsize=14)
    plt.xlabel('Месяц', fontsize=12)
    plt.ylabel('Выручка', fontsize=12)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    print(f"📊 Месяц с максимальными продажами: {monthly_sales.idxmax()} ({monthly_sales.max():,.2f})")
    print(f"📊 Месяц с минимальными продажами: {monthly_sales.idxmin()} ({monthly_sales.min():,.2f})")
```

```python
if discount_col and discount_col in df.columns:
    print("\n" + "="*50)
    print("📊 РАСПРЕДЕЛЕНИЕ СКИДОК")
    print("="*50)

    plt.figure(figsize=(10, 5))
    df[discount_col].hist(bins=20, edgecolor='black', color='skyblue')
    plt.title('Распределение скидок', fontsize=14)
    plt.xlabel('Скидка', fontsize=12)
    plt.ylabel('Количество заказов', fontsize=12)
    plt.grid(False)
    plt.show()

    print(f"📊 Средняя скидка: {df[discount_col].mean():.2f}")
    print(f"📊 Максимальная скидка: {df[discount_col].max():.2f}")
```

```python
if quantity_col and quantity_col in df.columns:
    print("\n" + "="*50)
    print("📊 РАСПРЕДЕЛЕНИЕ КОЛИЧЕСТВА ТОВАРОВ")
    print("="*50)

    plt.figure(figsize=(10, 5))
    df[quantity_col].hist(bins=20, edgecolor='black', color='lightgreen')
    plt.title('Распределение количества товаров в заказе', fontsize=14)
    plt.xlabel('Количество', fontsize=12)
    plt.ylabel('Количество заказов', fontsize=12)
    plt.grid(False)
    plt.show()

    print(f"📊 Среднее количество: {df[quantity_col].mean():.2f}")
    print(f"📊 Максимальное количество: {df[quantity_col].max()}")
```

```python
print("\n" + "="*50)
print("📊 ПРОДАЖИ ПО КАТЕГОРИЯМ (ИСПРАВЛЕННЫЙ ГРАФИК)")
print("="*50)

# 1. Ищем колонку с категорией
category_col = None
for col in df.columns:
    if 'category' in col.lower() and 'sub' not in col.lower():
        category_col = col
        break

# 2. Ищем колонку с выручкой
sales_col = None
for col in df.columns:
    if 'sales' in col.lower():
        sales_col = col
        break

# 3. Строим КРАСИВЫЙ график
if category_col and sales_col:
    # Считаем данные
    category_sales = df.groupby(category_col)[sales_col].sum().sort_values(ascending=False)
    print(category_sales)

    # Создаем график БОЛЬШОГО размера
    plt.figure(figsize=(12, 6))  # ← Шире и выше

    # Строим столбцы с красивым цветом
    bars = plt.bar(category_sales.index, category_sales.values, color='skyblue', edgecolor='black')

    # Подписываем значения на столбцах
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 500,  # ← +500 чтобы текст был над столбцом
                 f'{int(height)}', ha='center', va='bottom', fontsize=10)

    # Настройки заголовков
    plt.title('Продажи по категориям', fontsize=16, fontweight='bold')
    plt.xlabel('Категория', fontsize=14)
    plt.ylabel('Выручка', fontsize=14)

    # *** ЭТО САМОЕ ВАЖНОЕ: наклоняем подписи, чтобы они влезли ***
    plt.xticks(rotation=45, ha='right', fontsize=12)

    # Добавляем сетку для удобства
    plt.grid(axis='y', linestyle='--', alpha=0.3)

    # Автоматически подгоняем размер, чтобы ничего не обрезалось
    plt.tight_layout()

    # Показываем график
    plt.show()

    print(f"🏆 Самая популярная категория: {category_sales.index[0]}")

# 4. Строим ТОП-10 подкатегорий (тоже с большим графиком)
subcategory_col = None
for col in df.columns:
    if 'sub' in col.lower() and 'category' in col.lower():
        subcategory_col = col
        break

if subcategory_col and sales_col:
    subcategory_sales = df.groupby(subcategory_col)[sales_col].sum().sort_values(ascending=False).head(10)

    print("\n" + "="*50)
    print("📊 ТОП-10 ПОДКАТЕГОРИЙ ПО ВЫРУЧКЕ")
    print("="*50)
    print(subcategory_sales)

    # Строим график для подкатегорий
    plt.figure(figsize=(14, 7))  # Еще шире, так как названий больше

    bars = plt.bar(subcategory_sales.index, subcategory_sales.values, color='lightgreen', edgecolor='black')

    # Подписываем значения
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 300,
                 f'{int(height)}', ha='center', va='bottom', fontsize=9)

    plt.title('Топ-10 подкатегорий по выручке', fontsize=16, fontweight='bold')
    plt.xlabel('Подкатегория', fontsize=14)
    plt.ylabel('Выручка', fontsize=14)

    # *** НАКЛОН ДЛЯ ПОДКАТЕГОРИЙ ***
    plt.xticks(rotation=45, ha='right', fontsize=11)

    plt.grid(axis='y', linestyle='--', alpha=0.3)
    plt.tight_layout()
    plt.show()

else:
    print("❌ Колонка с подкатегорией не найдена")
```

```python
print("\n" + "="*50)
print("📊 АВТОМАТИЧЕСКИЙ ИТОГ")
print("="*50)

if order_col:
    print(f"Заказов: {df[order_col].nunique()}")
if customer_col:
    print(f"Клиентов: {df[customer_col].nunique()}")
if sales_col:
    print(f"Выручка: {df[sales_col].sum():,.2f}")
if category_col and sales_col:
    print(f"Топ-категория: {df.groupby(category_col)[sales_col].sum().idxmax()}")
if product_col and sales_col:
    print(f"Топ-товар: {df.groupby(product_col)[sales_col].sum().idxmax()}")
```

```python
print("\n" + "="*50)
print("ИТОГИ АНАЛИЗА AMAZON")
print("="*50)

print("\n1. ОБЩАЯ СТАТИСТИКА")
if 'Order ID' in df.columns:
    print(f"   Всего заказов: {df['Order ID'].nunique()}")
elif 'order_id' in df.columns:
    print(f"   Всего заказов: {df['order_id'].nunique()}")
else:
    print("   Всего заказов: (не найдено)")

if 'Customer ID' in df.columns:
    print(f"   Всего клиентов: {df['Customer ID'].nunique()}")
elif 'customer_id' in df.columns:
    print(f"   Всего клиентов: {df['customer_id'].nunique()}")
else:
    print("   Всего клиентов: (не найдено)")

if sales_col:
    print(f"   Общая выручка: {df[sales_col].sum():,.2f}")

print("\n2. КАТЕГОРИИ")
if category_col and sales_col:
    cat_sales = df.groupby(category_col)[sales_col].sum().sort_values(ascending=False)
    for i, (cat, sales) in enumerate(cat_sales.head(3).items(), 1):
        print(f"   {i}. {cat}: {sales:,.2f}")

print("\n3. ТОП-3 ПОДКАТЕГОРИИ")
if subcategory_col and sales_col:
    sub_sales = df.groupby(subcategory_col)[sales_col].sum().sort_values(ascending=False)
    for i, (sub, sales) in enumerate(sub_sales.head(3).items(), 1):
        print(f"   {i}. {sub}: {sales:,.2f}")

print("\n4. ДИНАМИКА")
if date_col and sales_col:
    df['month'] = pd.to_datetime(df[date_col], errors='coerce').dt.to_period('M')
    monthly = df.groupby('month')[sales_col].sum()
    print(f"   Пик продаж: {monthly.idxmax()} ({monthly.max():,.2f})")
    print(f"   Спад продаж: {monthly.idxmin()} ({monthly.min():,.2f})")

print("\n5. РЕКОМЕНДАЦИИ")
print("   1. Усилить маркетинг на категорию Technology")
print("   2. Развивать подкатегорию Phones")
print("   3. Запустить акции в месяц спада")
```
